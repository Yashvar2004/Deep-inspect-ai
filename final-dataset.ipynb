```python
import pandas as pd
import re

# Sample DataFrame
data = {'Info': ['[tcp spurious retransmission] 443  >  53661 [psh, ack] seq=1 ack=1 win=10 len=125']}
df = pd.DataFrame(data)

# Extract specific numeric value (e.g., 'len=125')
def extract_specific_numeric(text):
    match = re.search(r'len=(\d+)', text)
    if match:
        return float(match.group(1))
    return None

# Apply extraction
df['Info_numeric'] = df['Info'].apply(extract_specific_numeric)

# Handle NaN values by replacing them with 0
df['Info_numeric'] = df['Info_numeric'].fillna(0)

print(df)
```

```python
import pandas as pd

# Sample DataFrame with mixed data types
data = {
    'Source': ['A', 'B', 'A'],
    'Value1': ['10', '20', '30'],  # Should be numeric
    'Value2': ['5', '15', '25']    # Should be numeric
}
df = pd.DataFrame(data)

# Convert columns to numeric types
df['Value1'] = pd.to_numeric(df['Value1'], errors='coerce')
df['Value2'] = pd.to_numeric(df['Value2'], errors='coerce')

# Verify column types
print(df.dtypes)

# Group by 'Source' and compute mean for numeric columns
df_grouped = df.groupby('Source').mean()
print(df_grouped)
```
